## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

What I can say: which pages show observable signs of declining search visibility, thin content, or under-performance relative to their traffic — and a directional, decision-support ranking of which pages are worth reviewing first, given limited reviewer time.

What I can't say: that refreshing a flagged page will cause a recovery — proving that requires a controlled experiment, not observational data. I also can't claim anything about how Google's actual ranking algorithm works, or that a low-ranked page is "broken" — only that it's a lower priority given current evidence.

A note on the label itself: trend_direction (and therefore is_declining_label) is a defined rule, not a genuinely observed future outcome — it labels a page "down" if impressions fell more than 20% comparing the last 30 days to the prior 30 days (confirmed in docs/data-dictionary.md). Both windows have already happened by the time the label is assigned, so this is a proxy for decline, not a forecast. trend_direction and trend_pct will never be used as model features — only as the label — to avoid leakage.
